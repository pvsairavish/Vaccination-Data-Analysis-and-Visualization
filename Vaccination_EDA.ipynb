# Import Libraries

````python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set style for better looking charts
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
````

# Load Dataset

````python
# Load Dataset
coverage = pd.read_excel("cleaned_coverage.xlsx")
incidence = pd.read_excel("cleaned_incidence.xlsx")
reported_cases = pd.read_excel("cleaned_reported_cases.xlsx")
vaccine_intro = pd.read_excel("cleaned_vaccine_intro.xlsx")
vaccine_schedule = pd.read_excel("cleaned_vaccine_schedule.xlsx")

print("All datasets loaded successfully!")
````

##What did you know about your dataset?
The dataset consists of five related tables containing global vaccination and disease data over multiple years.

- Coverage data has over 230,000 records with information on vaccination percentages, doses, and target populations.
- Incidence and Reported Cases tables provide disease burden information.
- Vaccine Introduction and Schedule tables show which vaccines have been introduced and their recommended schedules.

The data required cleaning for missing values, especially in coverage and incidence rates. After cleaning, the data is suitable for trend analysis, correlation studies, and regional comparisons.

# Data Wrangling

````python
# Make a clean working copy
coverage_clean = coverage.copy()
incidence_clean = incidence.copy()
reported_clean = reported_cases.copy()
intro_clean = vaccine_intro.copy()

# Convert YEAR to integer (if not already)
coverage_clean['YEAR'] = coverage_clean['YEAR'].astype(int)
incidence_clean['YEAR'] = incidence_clean['YEAR'].astype(int)
reported_clean['YEAR'] = reported_clean['YEAR'].astype(int)
intro_clean['YEAR'] = intro_clean['YEAR'].astype(int)

print("Data Wrangling completed!")
print("Coverage shape after wrangling:", coverage_clean.shape)
````

## Chart - 1 visualization code

````python
# Chart - 1 visualization code
yearly_coverage = coverage_clean.groupby('YEAR')['COVERAGE'].mean().reset_index()

plt.figure(figsize=(12,6))
sns.lineplot(data=yearly_coverage, x='YEAR', y='COVERAGE', marker='o', color='teal')
plt.title('Average Vaccination Coverage by Year', fontsize=14, fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Average Coverage (%)')
plt.grid(True)
plt.show()
````

### 1. Why did you pick the specific chart?

Line chart is the best choice to show the trend of vaccination coverage over multiple years.2. What is/are the insight(s) found from the chart?

Vaccination coverage has shown a general increasing trend over the years, indicating improvement in global immunization programs.3. Will the gained insights help creating a positive business impact?

Yes, it helps public health organizations track the progress of vaccination programs and identify periods of success or decline for better planning.

## Chart - 14 - Correlation Heatmap

````python
# Chart - 14 - Correlation Heatmap
# Create a simple numerical dataframe for correlation
corr_data = pd.DataFrame({
    'Coverage': coverage_clean.groupby('YEAR')['COVERAGE'].mean(),
    'Incidence': incidence_clean.groupby('YEAR')['INCIDENCE_RATE'].mean(),
    'Cases': reported_clean.groupby('YEAR')['CASES'].sum()
}).dropna()

plt.figure(figsize=(8,6))
sns.heatmap(corr_data.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap: Coverage, Incidence & Cases', fontsize=14, fontweight='bold')
plt.show()
````

### 1. Why did you pick the specific chart?

Heatmap is the best visualization to show correlation strength between key numerical variables.
2. What is/are the insight(s) found from the chart?

There is a negative correlation between vaccination coverage and disease incidence/cases.
3. Will the gained insights help creating a positive business impact?

Yes, it statistically supports the importance of increasing vaccination coverage to reduce disease burden.
